```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

# Load the starter dataset
df = pd.read_csv("/content/content_refresh_anonymized (1).csv")

# Basic dataset size
print("Number of pages:", len(df))

# Pages with meaningful visibility but zero clicks
zero_click_pages = df[
    (df["impressions_90d"] >= 100) &
    (df["clicks_90d"] == 0)
]

print(
    "Pages with 100+ impressions and 0 clicks:",
    len(zero_click_pages)
)

print(
    "Percentage of all pages:",
    round(len(zero_click_pages) / len(df) * 100, 2),
    "%"
)

# Average CTR
print(
    "Average CTR:",
    round(df["ctr"].mean(), 4)
)
```
